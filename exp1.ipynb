```python
from collections import deque

def bfs(graph,start):
  visited=set()
  queue=deque([start])
  visited.add(start)
  print("BFS Traversal",end="")

  while queue:
    vertex=queue.popleft()
    print(vertex,end=" ") # Added space for better output formatting

    # Check if the vertex popped from the queue is a valid key in the graph
    if vertex not in graph:
      print(f"(Warning: Vertex '{vertex}' not found in graph, skipping its neighbors.)", end=" ")
      continue # Skip to the next item in the queue if this vertex is invalid

    for neighbor in graph[vertex]:
        if neighbor not in visited:
          visited.add(neighbor)
          queue.append(neighbor)

graph={}
while True:
    try:
        n=int(input("enter the number of vertex: ")) # Added space for better input prompt
        break
    except ValueError:
        print("Invalid input. Please enter an integer for the number of vertices.")
for i in range(n):
          vertex=input(f"enter vertex {i+1}:") # Used f-string for proper indexing
          # Clarified input prompt for neighbors
          neighbors=input(f"enter adjacent vertices of {vertex} (space separated, use existing vertex names): ").split()
          graph[vertex]=neighbors

print("\nGraph:",graph) # Moved print graph outside loop
start=input("enter the starting vertex: ") # Moved start input outside loop
bfs(graph,start)
```
